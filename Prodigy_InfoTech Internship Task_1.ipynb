# Bar Chart of a Categorical Variable (Gender Ratio)

```python
avg_men=census_dataset['Men'].sum()
avg_women=census_dataset['Women'].sum()
```

```python
# plot a bar graph to compare men and women population
plt.bar(['Men','Women'],[avg_men,avg_women],color=['blue','pink'])
plt.title('Total Men vs Women in US Counties (ACS 2017)')
plt.ylabel('Population')
plt.show()
```

# Histogram of a Continuous Variable (Median Household Income)

```python
# Histogram to show the Distribution of Median Household Income (USD)
plt.hist(census_dataset['Income'],bins=30,color='red',edgecolor='black')
plt.title('Distribution of Median Household Income (USD)')
plt.xlabel('Income ($)')
plt.ylabel('Number of Counties')
plt.show()
```

# Histogram to represent Poverty Percentage

```python
# Histogram to show the Distribution of Poverty Rate across Counties
plt.hist(census_dataset['Poverty'],bins=20,color='blue',edgecolor='black')
plt.title('Distribution of Poverty Rate across Counties')
plt.xlabel('Poverty Rate (%)')
plt.ylabel('Number of Counties')
plt.show()
```
